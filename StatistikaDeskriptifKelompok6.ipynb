```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import plotly.graph_objects as go
from IPython.display import display, HTML

# ============================================================
# VISUALISASI DATA BENCHMARK CS2 & STATISTIKA DESKRIPTIF
# Tema: Cyberpunk / Gaming Amber (Kuning & Hitam)
# Dataset: 3TIA_06_DataBenchmarkCS2.xlsx
# ============================================================

# ------------------------------------------------------------
# 1. PALET WARNA KHUSUS SETTING GRAFIK (TANPA WARNA HITAM)
# ------------------------------------------------------------
WARNA_GRAFIK = {
    "Low":       "#FACC15",  # Kuning Neon Terang
    "Medium":    "#F59E0B",  # Kuning Emas / Amber
    "High":      "#EA580C",  # Oranye Terang
    "Very High": "#EF4444"   # Merah Terang / Crimson
}

TEXT_GRAFIK = {
    "Low": "#000000",
    "Medium": "#000000",
    "High": "#ffffff",
    "Very High": "#ffffff"
}

# ------------------------------------------------------------
# 2. STYLING CSS INTERAKTIF (DARK THEME & TEKS TABEL CENTER)
# ------------------------------------------------------------
display(HTML("""
<style>
/* ============================================================
   VISUAL DASHBOARD — TEMA GAMING AMBER
   Fokus: rapi, jelas, tidak berlebihan
   ============================================================ */
:root {
    --bg-main: #0b0b0d;
    --bg-card: #151518;
    --bg-card-2: #111114;
    --border: #2b2b30;
    --text: #ededee;
    --muted: #a7a7ad;
    --accent: #facc15;
}

.judul {
    background: linear-gradient(135deg, #1b1b1f 0%, #0d0d0f 100%);
    color: var(--text) !important;
    padding: 24px 28px;
    border: 1px solid var(--border);
    border-top: 4px solid var(--accent);
    border-radius: 12px;
    margin: 4px 0 26px 0;
    box-shadow: 0 8px 24px rgba(0,0,0,.22);
    font-family: -apple-system, BlinkMacSystemFont, "Segoe UI", Roboto, sans-serif;
}

.judul h1 {
    margin: 0;
    font-size: 23px;
    line-height: 1.3;
    font-weight: 800;
    color: var(--accent) !important;
    letter-spacing: .4px;
}

.judul p {
    margin: 8px 0 0 0;
    font-size: 13.5px;
    line-height: 1.55;
    color: var(--muted) !important;
}

.bagian {
    background: #18181c;
    color: var(--accent) !important;
    padding: 11px 16px;
    border: 1px solid var(--border);
    border-left: 4px solid var(--accent);
    border-radius: 8px;
    margin: 28px 0 12px 0 !important;
    font-size: 14px;
    font-weight: 750;
    letter-spacing: .25px;
    font-family: -apple-system, BlinkMacSystemFont, "Segoe UI", Roboto, sans-serif;
}

.info {
    background: var(--bg-card-2) !important;
    color: #dddddf !important;
    border: 1px solid var(--border) !important;
    border-left: 4px solid #eab308 !important;
    padding: 12px 16px;
    border-radius: 8px;
    margin: 10px 0 14px 0 !important;
    font-size: 13px;
    line-height: 1.6;
    font-family: -apple-system, BlinkMacSystemFont, "Segoe UI", Roboto, sans-serif;
}

.data-panel,
.chart-container,
.control-panel {
    background: var(--bg-card);
    border: 1px solid var(--border);
    border-radius: 10px;
}

.data-panel {
    overflow-x: auto;
    margin: 12px 0;
}

table.tabel,
table.tabel-deskriptif {
    width: 100% !important;
    border-collapse: collapse !important;
    background: var(--bg-card) !important;
    color: var(--text) !important;
    font-family: -apple-system, BlinkMacSystemFont, "Segoe UI", Roboto, sans-serif;
    margin: 0 !important;
}

table.tabel {
    min-width: 860px;
    font-size: 12.5px !important;
}

table.tabel-deskriptif {
    font-size: 13px !important;
    border: 1px solid var(--border) !important;
    border-radius: 10px;
    overflow: hidden;
}

table.tabel th,
table.tabel-deskriptif th {
    background: #1b1b20 !important;
    color: var(--accent) !important;
    padding: 11px 12px !important;
    text-align: center !important;
    font-weight: 750 !important;
    border: 1px solid var(--border) !important;
    white-space: nowrap;
}

table.tabel td,
table.tabel-deskriptif td {
    background: #131317 !important;
    color: #dddddf !important;
    padding: 9px 11px !important;
    text-align: center !important;
    border: 1px solid #26262b !important;
}

table.tabel tbody tr:nth-child(even) td,
table.tabel-deskriptif tbody tr:nth-child(even) td {
    background: #16161a !important;
}

table.tabel tbody tr:hover td,
table.tabel-deskriptif tbody tr:hover td {
    background: #202025 !important;
}

table.tabel th.sortable {
    user-select: none;
}

table.tabel th.sortable:hover {
    background: #26262c !important;
    color: #ffffff !important;
    cursor: pointer !important;
}

table.tabel-deskriptif td.nilai {
    color: var(--accent) !important;
    font-weight: 750 !important;
}

.control-panel {
    padding: 14px 16px;
    margin: 12px 0;
}

.box-filter {
    display: flex;
    align-items: center;
    flex-wrap: wrap;
    gap: 10px;
    margin: 0;
    font-family: -apple-system, BlinkMacSystemFont, "Segoe UI", Roboto, sans-serif;
}

.label-filter {
    color: var(--accent);
    font-weight: 700;
    font-size: 13px;
}

.select-filter {
    padding: 8px 12px;
    border: 1px solid #404047;
    border-radius: 7px;
    background: #111114;
    color: #ffffff;
    font-size: 13px;
    font-weight: 600;
    outline: none;
    cursor: pointer;
}

.select-filter:hover,
.select-filter:focus {
    border-color: var(--accent);
}

.radio-group-container {
    padding: 14px 16px;
    margin: 12px 0;
    background: #151518;
    border: 1px solid var(--border);
    border-radius: 10px;
    font-family: -apple-system, BlinkMacSystemFont, "Segoe UI", Roboto, sans-serif;
}

.radio-group-title {
    display: block;
    margin-bottom: 10px;
    color: var(--accent);
    font-size: 13px;
    font-weight: 750;
}

.radio-box-group {
    display: flex;
    flex-wrap: wrap;
    gap: 9px;
}

.radio-box-item {
    display: flex;
    align-items: center;
    gap: 8px;
    padding: 8px 12px;
    border: 1px solid #3b3b42;
    border-radius: 7px;
    background: #111114;
    color: #d6d6da;
    font-size: 12.5px;
    font-weight: 650;
    cursor: pointer;
    user-select: none;
    transition: border-color .15s ease, background .15s ease, color .15s ease;
}

.radio-box-item:hover {
    border-color: var(--accent);
    color: #ffffff;
}

.radio-box-item input[type="radio"] {
    width: 15px;
    height: 15px;
    accent-color: var(--accent);
    cursor: pointer;
}

.radio-box-item.active {
    border-color: var(--accent) !important;
    background: rgba(250, 204, 21, .08) !important;
    color: var(--accent) !important;
}

.chart-container {
    padding: 6px;
    margin-top: 10px;
    overflow: hidden;
}

/* Lebar output notebook lebih nyaman */
.jp-OutputArea-output,
.output_subarea {
    max-width: 100% !important;
}
</style>
"""))

# Header Utama
display(HTML("""
<div class="judul">
    <h1>BENCHMARK PERFORMANCE CS2 — ANALISIS STATISTIK</h1>
    <p>Visualisasi Performa Frame Per Second (FPS) Berdasarkan Quality Preset (Low, Medium, High, Very High)</p>
</div>
"""))

# ------------------------------------------------------------
# 3. LOAD DATA BENCHMARK
# ------------------------------------------------------------
# Pilih file dataset dari komputer (Google Colab)
from google.colab import files
import os

uploaded = files.upload()

# Ambil file pertama yang didukung
supported_extensions = (".xlsx", ".xls", ".xlsm", ".csv", ".tsv")
FILE_NAME = next((name for name in uploaded if name.lower().endswith(supported_extensions)), None)

if FILE_NAME is None:
    raise ValueError("File tidak didukung. Gunakan .xlsx, .xls, .xlsm, .csv, atau .tsv.")

SHEET_NAME = None

display(HTML(f"""
<div class="bagian">1. LOAD DATA BENCHMARK</div>
<div class="info">
    Dataset yang dipilih: <b style="color: #facc15;">{FILE_NAME}</b>
</div>
"""))

# Baca Data
extension = os.path.splitext(FILE_NAME)[1].lower()

if extension in (".xlsx", ".xls", ".xlsm"):
    excel_file = pd.ExcelFile(FILE_NAME)
    sheet_names = excel_file.sheet_names

    if "Dataset" in sheet_names:
        SHEET_NAME = "Dataset"
    else:
        SHEET_NAME = sheet_names[0]

    display(HTML(f"""
    <div class="info">Sheet yang digunakan: <b style="color: #facc15;">{SHEET_NAME}</b>
    <br>Sheet tersedia: {', '.join(sheet_names)}</div>
    """))

    df = pd.read_excel(excel_file, sheet_name=SHEET_NAME)
else:
    if extension == ".tsv":
        df = pd.read_csv(FILE_NAME, sep="\t")
    else:
        df = pd.read_csv(FILE_NAME)

df.columns = ['No', 'Jenis_Grafik', 'MinimumFramePerSeconds', 'MaximumFramePerSeconds', 'AverageFramePerSeconds', '0.1%FramePerSeconds']
df.columns = [c.strip() for c in df.columns]

# Pewarnaan Tabel HTML
def mewarnai_tabel_html(df_in, table_id="", enable_sort=False):
    html_out = f"<table class='tabel' id='{table_id}'><thead><tr>"
    for idx, col in enumerate(df_in.columns):
        if enable_sort:
            html_out += f"<th class='sortable' onclick='sortTable(\"{table_id}\", {idx})' title='Klik untuk mengurutkan'>{col} ⇳</th>"
        else:
            html_out += f"<th>{col}</th>"
    html_out += "</tr></thead><tbody>"

    for _, row in df_in.iterrows():
        html_out += "<tr>"
        for col in df_in.columns:
            val = row[col]
            if col == "Jenis_Grafik" and str(val) in WARNA_GRAFIK:
                bg = WARNA_GRAFIK[str(val)]
                txt = TEXT_GRAFIK.get(str(val), "#ffffff")
                html_out += f"<td style='background-color: {bg} !important; color: {txt} !important; font-weight: bold; text-align: center !important;'>{val}</td>"
            else:
                html_out += f"<td style='text-align: center !important;'>{val if pd.notnull(val) else ''}</td>"
        html_out += "</tr>"
    html_out += "</tbody></table>"
    return html_out

# ------------------------------------------------------------
# 4. TABEL DATA OLAHAN (DENGAN FILTER & SORTING)
# ------------------------------------------------------------
display(HTML("""
<div class="bagian">2. DATA HASIL BENCHMARK CS2</div>
<div class="info">
    • <b style="color: #facc15;">Filter:</b> Pilih jenis grafik untuk menyaring baris data.<br>
    • <b style="color: #facc15;">Sorting:</b> Klik header kolom mana saja untuk mengurutkan nilai FPS.
</div>

<div class="box-filter">
    <span class="label-filter">Filter Jenis Grafik:</span>
    <select id="filterSelectGrafik" class="select-filter" onchange="filterBenchmarkByGrafik()">
        <option value="ALL">Semua Jenis Grafik (All)</option>
        <option value="LOW">Low</option>
        <option value="MEDIUM">Medium</option>
        <option value="HIGH">High</option>
        <option value="VERY HIGH">Very High</option>
    </select>
</div>

<script>
function filterBenchmarkByGrafik() {
    var select = document.getElementById("filterSelectGrafik");
    var filterVal = select.value.toUpperCase();
    var table = document.getElementById("tabelBenchmarkCS2");
    if (!table) return;
    var tr = table.getElementsByTagName("tr");

    for (var i = 1; i < tr.length; i++) {
        var tdGrafik = tr[i].getElementsByTagName("td")[1];
        if (tdGrafik) {
            var txtValue = (tdGrafik.textContent || tdGrafik.innerText).trim().toUpperCase();
            if (filterVal === "ALL" || txtValue === filterVal) {
                tr[i].style.display = "";
            } else {
                tr[i].style.display = "none";
            }
        }
    }
}

var sortDirections = {};
function sortTable(tableId, colIndex) {
    var table = document.getElementById(tableId);
    var tbody = table.querySelector("tbody");
    var rows = Array.from(tbody.querySelectorAll("tr"));

    sortDirections[colIndex] = !sortDirections[colIndex];
    var asc = sortDirections[colIndex];

    rows.sort(function(a, b) {
        var cellA = a.children[colIndex].textContent.trim();
        var cellB = b.children[colIndex].textContent.trim();

        var numA = parseFloat(cellA);
        var numB = parseFloat(cellB);

        if (!isNaN(numA) && !isNaN(numB)) {
            return asc ? numA - numB : numB - numA;
        } else {
            return asc ? cellA.localeCompare(cellB) : cellB.localeCompare(cellA);
        }
    });

    rows.forEach(function(row) { tbody.appendChild(row); });

    var headers = table.querySelectorAll("th");
    headers.forEach(function(th, idx) {
        var baseText = th.getAttribute("data-name");
        if (!baseText) {
            baseText = th.textContent.replace(/[⇳▲▼]/g, "").trim();
            th.setAttribute("data-name", baseText);
        }
        th.innerHTML = (idx === colIndex) ? baseText + " " + (asc ? "▲" : "▼") : baseText + " ⇳";
    });
}
</script>
"""))

display(HTML(f"<div class='data-panel'>{mewarnai_tabel_html(df, table_id='tabelBenchmarkCS2', enable_sort=True)}</div>"))

# ------------------------------------------------------------
# 5. TABEL RINGKASAN RATA-RATA FPS PER JENIS GRAFIK
# ------------------------------------------------------------
display(HTML("""<div class="bagian">3. RINGKASAN RATA-RATA FPS PER PRESET GRAFIK</div>"""))

df_summary = df.groupby("Jenis_Grafik")[
    ["MinimumFramePerSeconds", "MaximumFramePerSeconds", "AverageFramePerSeconds", "0.1%FramePerSeconds"]
].mean().reset_index()

# Reorder kategori
order_preset = ["Low", "Medium", "High", "Very High"]
df_summary["Jenis_Grafik"] = pd.Categorical(df_summary["Jenis_Grafik"], categories=order_preset, ordered=True)
df_summary = df_summary.sort_values("Jenis_Grafik").round(2)

df_summary.columns = ["Jenis_Grafik", "Rata-rata Min FPS", "Rata-rata Max FPS", "Rata-rata Avg FPS", "Rata-rata 0.1% Low FPS"]
display(HTML(f"<div class='data-panel'>{mewarnai_tabel_html(df_summary)}</div>"))

# ------------------------------------------------------------
# 6. VISUALISASI INTERAKTIF DENGAN PEMILIHAN RADIO BOX
# ------------------------------------------------------------
display(HTML("""
<div class="bagian">4. VISUALISASI DATA INTERAKTIF</div>
<div class="info">
    Pilih jenis grafik menggunakan <b>Radio Box</b> di bawah ini untuk menampilkan visualisasi data secara interaktif.
</div>
"""))

# Format Layout Dark Plotly
layout_plotly_dark = dict(
    template="plotly_dark",
    paper_bgcolor="#151518",
    plot_bgcolor="#151518",
    font=dict(color="#e4e4e7", family="-apple-system, BlinkMacSystemFont, 'Segoe UI', Roboto, sans-serif"),
    title=dict(x=0.02, xanchor="left", font=dict(size=18)),
    xaxis=dict(gridcolor="#2b2b30", zerolinecolor="#2b2b30", showline=True, linecolor="#3a3a40"),
    yaxis=dict(gridcolor="#2b2b30", zerolinecolor="#2b2b30", showline=True, linecolor="#3a3a40"),
    height=450,
    margin=dict(l=55, r=30, t=65, b=50),
    legend=dict(orientation="h", yanchor="bottom", y=1.02, xanchor="left", x=0)
)

# Grafik 1: Line Chart - Average FPS
fig_line_fps = px.line(
    df, x="No", y="AverageFramePerSeconds", color="Jenis_Grafik",
    color_discrete_map=WARNA_GRAFIK, markers=True,
    title="<b>Grafik Garis: Sebaran Average FPS Seluruh Pengujian CS2</b>"
)
fig_line_fps.update_layout(**layout_plotly_dark)
fig_line_fps.update_layout(xaxis_title="Nomor Pengujian", yaxis_title="Average FPS")

# Grafik 2: Bar Chart - Average FPS
fig_bar_fps = px.bar(
    df_summary, x="Jenis_Grafik", y="Rata-rata Avg FPS", color="Jenis_Grafik",
    color_discrete_map=WARNA_GRAFIK, text="Rata-rata Avg FPS",
    title="<b>Grafik Batang: Rata-Rata Avg FPS per Preset Grafik</b>"
)
fig_bar_fps.update_traces(textposition="outside")
fig_bar_fps.update_layout(**layout_plotly_dark)
fig_bar_fps.update_layout(xaxis_title="Jenis Grafik", yaxis_title="Avg FPS", showlegend=False)

# Grafik 3: Box Plot - Average FPS
fig_box_fps = px.box(
    df, x="Jenis_Grafik", y="AverageFramePerSeconds", color="Jenis_Grafik",
    color_discrete_map=WARNA_GRAFIK, points="all",
    title="<b>Box Plot: Distribusi & Pencilan (Outlier) Average FPS</b>"
)
fig_box_fps.update_layout(**layout_plotly_dark)
fig_box_fps.update_layout(xaxis_title="Jenis Grafik", yaxis_title="Average FPS")

# Grafik 1: Line Chart - 0.1% Low FPS
fig_line_low = px.line(
    df, x="No", y="0.1%FramePerSeconds", color="Jenis_Grafik",
    color_discrete_map=WARNA_GRAFIK, markers=True,
    title="<b>Grafik Garis: Sebaran 0.1% Low FPS Seluruh Pengujian CS2</b>"
)
fig_line_low.update_layout(**layout_plotly_dark)
fig_line_low.update_layout(xaxis_title="Nomor Pengujian", yaxis_title="0.1% Low FPS")

# Grafik 2: Bar Chart - 0.1% Low FPS
fig_bar_low = px.bar(
    df_summary, x="Jenis_Grafik", y="Rata-rata 0.1% Low FPS", color="Jenis_Grafik",
    color_discrete_map=WARNA_GRAFIK, text="Rata-rata 0.1% Low FPS",
    title="<b>Grafik Batang: Rata-Rata 0.1% Low FPS per Preset Grafik</b>"
)
fig_bar_low.update_traces(textposition="outside")
fig_bar_low.update_layout(**layout_plotly_dark)
fig_bar_low.update_layout(xaxis_title="Jenis Grafik", yaxis_title="0.1% Low FPS", showlegend=False)

# Grafik 3: Box Plot - 0.1% Low FPS
fig_box_low = px.box(
    df, x="Jenis_Grafik", y="0.1%FramePerSeconds", color="Jenis_Grafik",
    color_discrete_map=WARNA_GRAFIK, points="all",
    title="<b>Box Plot: Distribusi & Pencilan (Outlier) 0.1% Low FPS</b>"
)
fig_box_low.update_layout(**layout_plotly_dark)
fig_box_low.update_layout(xaxis_title="Jenis Grafik", yaxis_title="0.1% Low FPS")

html_line_fps = fig_line_fps.to_html(full_html=False, include_plotlyjs='cdn')
html_bar_fps = fig_bar_fps.to_html(full_html=False, include_plotlyjs=False)
html_box_fps = fig_box_fps.to_html(full_html=False, include_plotlyjs=False)
html_line_low = fig_line_low.to_html(full_html=False, include_plotlyjs=False)
html_bar_low = fig_bar_low.to_html(full_html=False, include_plotlyjs=False)
html_box_low = fig_box_low.to_html(full_html=False, include_plotlyjs=False)

# PEMILIHAN METRIK GRAFIK MENGGUNAKAN DROPDOWN
# Dan tetap menggunakan radio box untuk memilih jenis grafik.
display(HTML(f"""
<div class="radio-group-container">
    <div class="metric-filter-row">
        <span class="radio-group-title">Pilih Data yang Ditampilkan:</span>
        <select id="metricSelect" class="select-filter" onchange="switchMetric(this.value)">
            <option value="fps" selected>Average FPS</option>
            <option value="low">0.1% Low FPS</option>
        </select>
    </div>

    <span class="radio-group-title">Pilih Jenis Grafik:</span>
    <div class="radio-box-group">
        <label class="radio-box-item active" id="lbl-line">
            <input type="radio" name="radioChartType" value="line" checked onclick="switchChartRadio('line')">
            Grafik Garis (Line Chart)
        </label>
        <label class="radio-box-item" id="lbl-bar">
            <input type="radio" name="radioChartType" value="bar" onclick="switchChartRadio('bar')">
            Grafik Batang (Bar Chart)
        </label>
        <label class="radio-box-item" id="lbl-box">
            <input type="radio" name="radioChartType" value="box" onclick="switchChartRadio('box')">
            Box Plot (Distribusi Data)
        </label>
    </div>
</div>

<div id="container-line-fps" class="chart-container" style="display: block;">{html_line_fps}</div>
<div id="container-bar-fps" class="chart-container" style="display: none;">{html_bar_fps}</div>
<div id="container-box-fps" class="chart-container" style="display: none;">{html_box_fps}</div>
<div id="container-line-low" class="chart-container" style="display: none;">{html_line_low}</div>
<div id="container-bar-low" class="chart-container" style="display: none;">{html_bar_low}</div>
<div id="container-box-low" class="chart-container" style="display: none;">{html_box_low}</div>

<script>
var selectedMetric = 'fps';
var selectedChart = 'line';

function switchMetric(metric) {{
    selectedMetric = metric;
    updateChartDisplay();
}}

function switchChartRadio(type) {{
    selectedChart = type;
    document.getElementById("lbl-line").classList.remove("active");
    document.getElementById("lbl-bar").classList.remove("active");
    document.getElementById("lbl-box").classList.remove("active");

    if (type === 'line') document.getElementById("lbl-line").classList.add("active");
    if (type === 'bar') document.getElementById("lbl-bar").classList.add("active");
    if (type === 'box') document.getElementById("lbl-box").classList.add("active");

    updateChartDisplay();
}}

function updateChartDisplay() {{
    var chartTypes = ['line', 'bar', 'box'];
    var metrics = ['fps', 'low'];

    chartTypes.forEach(function(chart) {{
        metrics.forEach(function(metric) {{
            var container = document.getElementById("container-" + chart + "-" + metric);
            container.style.display = (chart === selectedChart && metric === selectedMetric) ? 'block' : 'none';
        }});
    }});

    window.dispatchEvent(new Event('resize'));
}}
</script>
"""))

# ------------------------------------------------------------
# 7. STATISTIKA DESKRIPTIF LENGKAP (TEKS DITENGAHKAN)
# ------------------------------------------------------------
display(HTML("""<div class="bagian">5. STATISTIKA DESKRIPTIF (AVERAGE FPS)</div>"""))

fps_data = df["AverageFramePerSeconds"]

stat_df = pd.DataFrame({
    "Ukuran Statistik": [
        "Jumlah Data (N)", "Rata-rata (Mean)", "Nilai Tengah (Median)",
        "Nilai Minimum", "Nilai Maksimum", "Kuartil 1 (Q1)", "Kuartil 3 (Q3)",
        "Interquartile Range (IQR)", "Varians", "Simpangan Baku (Std Dev)",
        "Kemencengan (Skewness)", "Keruncingan (Kurtosis)"
    ],
    "Nilai Kalkulasi": [
        len(fps_data),
        f"{fps_data.mean():.2f}",
        f"{fps_data.median():.2f}",
        f"{fps_data.min():.2f}",
        f"{fps_data.max():.2f}",
        f"{fps_data.quantile(0.25):.2f}",
        f"{fps_data.quantile(0.75):.2f}",
        f"{(fps_data.quantile(0.75) - fps_data.quantile(0.25)):.2f}",
        f"{fps_data.var():.2f}",
        f"{fps_data.std():.2f}",
        f"{fps_data.skew():.2f}",
        f"{fps_data.kurt():.2f}"
    ]
})

html_stat = "<table class='tabel-deskriptif'><thead><tr><th style='text-align:center;'>Ukuran Statistik</th><th style='text-align:center;'>Nilai Kalkulasi</th></tr></thead><tbody>"
for _, r in stat_df.iterrows():
    html_stat += f"<tr><td style='text-align:center;'>{r['Ukuran Statistik']}</td><td class='nilai' style='text-align:center;'>{r['Nilai Kalkulasi']}</td></tr>"
html_stat += "</tbody></table>"

display(HTML(html_stat))
```
